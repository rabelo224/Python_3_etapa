# Projeto Final – Análise da População

Nesta atividade serão utilizados os dados do Censo Demográfico de 2022 para comparar a população de 2010 e 2022. Primeiro será feita a análise por estado e depois por município.

```python
import pandas as pd

arquivo = "CD2022_Populacao_2010_Compatibilizada_20231222.xlsx"

dados = pd.read_excel(arquivo, sheet_name="Municípios", header=2)
dados = dados.dropna(subset=["UF", "NOME DO MUNICÍPIO"])

dados.head()
```

## Preparação dos dados

As colunas de população serão convertidas para números para que seja possível fazer os cálculos.

```python
coluna_2010 = "População 2010 (Alterações de Limites até 2022)1"
coluna_2022 = "População Censo 2022"

dados[coluna_2010] = pd.to_numeric(dados[coluna_2010], errors="coerce")
dados[coluna_2022] = pd.to_numeric(dados[coluna_2022], errors="coerce")

dados["DIFERENCA"] = dados[coluna_2022] - dados[coluna_2010]

dados[["UF", "NOME DO MUNICÍPIO", coluna_2010, coluna_2022, "DIFERENCA"]].head()
```

## 1. População por estado

```python
estado = dados.groupby("UF", as_index=False)[[coluna_2010, coluna_2022]].sum()

estado["DIFERENCA"] = estado[coluna_2022] - estado[coluna_2010]
estado = estado.sort_values("DIFERENCA", ascending=False).reset_index(drop=True)

estado
```

## 2. Salvar o resultado dos estados

A tabela é salva em formato CSV para facilitar a utilização dos resultados em outros programas.

```python
estado.to_csv(
    "populacao_por_estado.csv",
    sep=";",
    index=False,
    encoding="utf-8-sig"
)

print("Tabela por estado salva com sucesso.")
```

## 3. População por município

```python
municipio = dados[[
    "UF",
    "COD. UF",
    "COD. MUNIC",
    "NOME DO MUNICÍPIO",
    coluna_2010,
    coluna_2022,
    "DIFERENCA"
]].copy()

municipio = municipio.sort_values("DIFERENCA", ascending=False)
municipio = municipio.reset_index(drop=True)

municipio.head(20)
```

## 4. Salvar o resultado dos municípios

```python
municipio.to_csv(
    "populacao_por_municipio.csv",
    sep=";",
    index=False,
    encoding="utf-8-sig"
)

print("Tabela por município salva com sucesso.")
```

## Conferência dos resultados

Abaixo aparecem os estados e municípios com as maiores diferenças de população.

```python
print("5 estados com maior diferença:")
display(estado.head(5))

print("10 municípios com maior diferença:")
display(municipio.head(10))
```
